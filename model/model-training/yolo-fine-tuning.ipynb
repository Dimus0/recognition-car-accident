```python
from ultralytics import YOLO
```

```python
model = YOLO("yolo11n.pt")
```

```python
class_names = ['car','bigcar']
```

```python
import yaml

yaml_data = {
    'train' : '/kaggle/input/datasets/puhachevskyi/a9-dataset/dataset/images/train',
    'val': '/kaggle/input/datasets/puhachevskyi/a9-dataset/dataset/images/val',

    'nc': 2,

    'names': ['car','bigcar']
}

yaml_path = '/kaggle/working/data.yaml'

with open(yaml_path, 'w') as f:
    yaml.dump(yaml_data, f, default_flow_style=False)

print(f"Файл data.yaml успішно створено за шляхом: {yaml_path}")

# Виводимо вміст файлу для перевірки
with open(yaml_path, 'r') as f:
    print(f.read())
```

```python
import torch
print(torch.cuda.is_available())
print(torch.cuda.get_device_name(0))
print(torch.version.cuda)
```

```python
import gc
gc.collect()
torch.cuda.empty_cache()
print(f"Allocated: {torch.cuda.memory_allocated() / 1024**3:.2f} GB")
print(f"Reserved:  {torch.cuda.memory_reserved() / 1024**3:.2f} GB")
```

```python
model.train(
    data="/kaggle/working/data.yaml",  # Шлях до файлу з описом датасету
    epochs=50,         # Кількість епох для навчання
    batch=32,         # Розмір батчу
    imgsz=640,
    patience=3,
    lr0=0.001,
    lrf=0.01,
    freeze=10,
    device=[0,1],
    name='yolov11',
    pretrained=True,
    exist_ok=True
)
```

```python
metrics = model.eval()
print(metrics)
```
